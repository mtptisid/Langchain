
## Part 10 — Streaming Responses

By default, `.invoke()` waits for the entire response before returning it.
For long answers this means staring at a blank screen for several seconds.

**Streaming** sends tokens as they are generated — the response appears
word-by-word, just like ChatGPT's interface.

````python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=0)

prompt = ChatPromptTemplate.from_template(
    "Explain {topic} in shorts for a beginner."
)
chain = prompt | llm | StrOutputParser()

# .stream() yields tokens as they arrive
for chunk in chain.stream({"topic": "how vector databases work"}):
    print(chunk, end="", flush=True)   # flush=True ensures immediate display
print()
````

**Streaming a RAG chain (same API):**

````python
rag_chain = prompt | llm | StrOutputParser()
# Works identically on any chain — just replace .invoke() with .stream()
for chunk in rag_chain.stream("What does NIST say about AI risk governance?"):
    print(chunk, end="", flush=True)
print()
````

**Async streaming — for web applications:**


````python
import asyncio
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser


llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=0)

prompt = ChatPromptTemplate.from_template(
    "Explain {topic} in shorts for a beginner."
)


rag_chain = prompt | llm | StrOutputParser()

async def stream_async(question: str):
    async for chunk in rag_chain.astream({"topic": question}):
        print(chunk, end="", flush=True)
    print()

await stream_async("transformer architecture")
````

**When to use which:**

| Method | Use when |
|---|---|
| `.invoke()` | Short responses, batch processing, testing |
| `.stream()` | User-facing apps — much better UX |
| `.astream()` | Async web frameworks (FastAPI, Starlette) |
| `.batch()` | Processing many inputs in parallel |

## Part 11 — Agents

### What Is an Agent?

A regular chain follows a fixed path: step 1 → step 2 → step 3.

An **agent** is dynamic. It decides what to do at each step based on what
it knows so far. It can call tools, read the results, call different tools,
and loop until it has enough information to answer.

```
User: "Who signed the EU AI Act and when?"
         │
         ▼
    LLM thinks: "I need to search for this."
         │
         ▼
    Calls: DuckDuckGo("EU AI Act signed date")
         │
         ▼
    Reads result → "Still need the exact date. Search Wikipedia."
         │
         ▼
    Calls: Wikipedia("EU Artificial Intelligence Act")
         │
         ▼
    Has enough info → generates final answer
```

This loop is called **ReAct** (Reason + Act).

### The LangChain Agent Version Problem

LangChain's agent imports have changed multiple times:

```py
# Older versions used:
from langchain.agents import initialize_agent, AgentType         # removed

# Newer versions use:
from langchain.agents import create_tool_calling_agent  
```

If you're on a version where neither exists, both imports throw `ImportError`.

**The fix: build the agent loop yourself using only `langchain_core.messages`
— which has been stable across all versions.**

### Version-Proof Agent (Recommended)

No `langchain.agents` import. Works on any LangChain version.

````python
!pip install -U ddgs
````

````python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_community.tools import WikipediaQueryRun, DuckDuckGoSearchRun
from langchain_community.utilities import WikipediaAPIWrapper
from langchain_core.messages import HumanMessage, ToolMessage

# ── Tools ──────────────────────────────────────────────────────────────────
wiki_tool   = WikipediaQueryRun(api_wrapper=WikipediaAPIWrapper(top_k_results=2))
search_tool = DuckDuckGoSearchRun()

# Map name → callable so we can invoke by name
tool_map = {
    "wikipedia":         wiki_tool,
    "duckduckgo_search": search_tool,
}

# ── LLM with tools bound ───────────────────────────────────────────────────
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=0)
llm_with_tools = llm.bind_tools(list(tool_map.values()))

# ── Agent loop ─────────────────────────────────────────────────────────────
def run_agent(question: str, max_iterations: int = 5) -> str:
    """
    Manual ReAct loop:
    1. Send messages to LLM
    2. If LLM returns tool_calls → execute tools, append results, repeat
    3. If LLM returns no tool_calls → it has the final answer, return it
    """
    messages = [HumanMessage(content=question)]
    print(f"\nQuestion: {question}")
    print("─" * 60)

    for i in range(max_iterations):
        response = llm_with_tools.invoke(messages)
        messages.append(response)

        # No tool calls → LLM has its final answer
        if not response.tool_calls:
            print(f"\nFinal Answer:\n{response.content}")
            return response.content

        # Execute each tool the LLM decided to use
        for tool_call in response.tool_calls:
            name  = tool_call["name"]
            args  = tool_call["args"]
            tc_id = tool_call["id"]

            print(f"\n[Step {i+1}] Tool: {name}")
            print(f"  Input:  {args}")

            tool_fn = tool_map.get(name)
            result  = tool_fn.invoke(args) if tool_fn else "Tool not found."

            print(f"  Result: {str(result)[:200]}...")

            # Return tool result to LLM so it can continue reasoning
            messages.append(ToolMessage(
                content=str(result),
                tool_call_id=tc_id
            ))

    return "Max iterations reached without a final answer."


# ── Run ────────────────────────────────────────────────────────────────────
result = run_agent("Who wrote the EU AI Act and when was it signed into law?")
````

### How `.bind_tools()` Works

When you call `llm.bind_tools(tools)`, LangChain automatically:

1. Reads each tool's name, description, and input schema
2. Formats them into the Gemini function-calling format
3. Sends them with every LLM call so Gemini knows what tools it has access to

Gemini then responds with either:
- `content` — a text answer (no tool needed)
- `tool_calls` — a list of tools it wants to call, with arguments

```python
# What a tool_call looks like in the response
tool_call = {
    "name": "duckduckgo_search",
    "args": {"query": "EU AI Act signed date"},
    "id":   "call_abc123"
}
```

The `id` is how you match each `ToolMessage` back to the correct call —
Gemini uses it to track which result belongs to which tool call.

### Build Your Own Custom Tool

Any Python function can become a tool with the `@tool` decorator.

````python
from langchain_core.tools import tool

@tool
def word_count(text: str) -> str:
    """Count the number of words in a piece of text. Input is the text string."""
    count = len(text.split())
    return f"{count} words"

@tool
def calculate(expression: str) -> str:
    """Evaluate a simple mathematical expression. Input is the expression string."""
    try:
        result = eval(expression, {"__builtins__": {}})
        return str(result)
    except Exception as e:
        return f"Error: {e}"

# Add to tool_map and the agent picks them up automatically
tool_map["word_count"] = word_count
tool_map["calculate"]  = calculate
llm_with_tools = llm.bind_tools(list(tool_map.values()))
````

**What makes a good tool description?**
The LLM reads the docstring to decide *when* to use the tool. Be specific:

```python
# Bad — too vague
"""Search for information."""

# Good — tells LLM exactly what it's for and what input to provide
"""Search Wikipedia for factual information about people, events, or concepts.
Input should be a specific search query string."""
```

### If You Have LangChain 0.1+ — Using create_tool_calling_agent

If your LangChain version supports it, the built-in approach is cleaner:

````python
from langchain.agents import AgentExecutor, create_tool_calling_agent
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a research assistant. Use tools to find accurate answers."),
    ("human", "{input}"),
    MessagesPlaceholder(variable_name="agent_scratchpad")
])

agent    = create_tool_calling_agent(llm, list(tool_map.values()), prompt)
executor = AgentExecutor(
    agent=agent,
    tools=list(tool_map.values()),
    verbose=True,
    max_iterations=5
)

result = executor.invoke({"input": "Who signed the EU AI Act and when?"})
print(result["output"])
````

**When to use which:**

| Approach | Requirement | Best for |
|---|---|---|
| Manual loop (our approach) | Any LangChain version | Learning, debugging, production control |
| `create_tool_calling_agent` | LangChain ≥ 0.1 | Cleaner code when version is guaranteed |
| `initialize_agent` | LangChain < 0.1 | Legacy code only |

## Full Agent Reference

````python
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage




# Define tools
@tool
def add(a: int, b: int) -> int:
    """Add two numbers."""
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers."""
    return a * b

# Bind tools
llm_with_tools = llm.bind_tools([add, multiply])

# Messages
messages = [HumanMessage(content="What is 5 + 7?")]

# Invoke model
response = llm_with_tools.invoke(messages)

print(response)

# Check tool calls
if response.tool_calls:
    for tc in response.tool_calls:
        name = tc["name"]
        args = tc["args"]
        tc_id = tc["id"]

        print("Tool:", name)
        print("Args:", args)
        print("ID:", tc_id)
````
